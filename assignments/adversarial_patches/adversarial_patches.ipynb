## Introduction

In this notebook, I explore the concept of **adversarial robustness** by creating an **adversarial patch**—a localized visual pattern that intentionally fools a neural network classifier. Using a pretrained **ResNet34** model trained on ImageNet, I will generate a patch that alters model predictions in a controlled and reproducible way.

The goal is to understand how small, targeted visual perturbations can drastically affect a model’s confidence and prediction, highlighting vulnerabilities in deep learning systems. Beyond the technical implementation, this assignment also challenges me to apply a **creative disguise or extension** to the patch, such as embedding it into a sticker, household object, or composite design.


## Goal in this Notebook

- Use a pretrained **ResNet34** (`torchvision.models.resnet34(weights="IMAGENET1K_V1")`)  
- Create an **adversarial patch** that intentionally manipulates classification results  
- Explore **creative extensions**, such as:
  - Disguising the patch as a real-world object (e.g., logo, clothing, or sticker)
  - Combining multiple patches to observe compound effect
  - Sending a hidden message using sequential patches
- Evaluate and visualize model predictions **before and after** patch application
- Present a **physical version** of the patch for in-class testing
